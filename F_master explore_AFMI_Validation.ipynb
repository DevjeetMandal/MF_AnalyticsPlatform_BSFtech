```python
master_codes = set(fund_master["amfi_code"])
```

```python
for name, df in [
    ("nav_history", nav_history),
    ("investor_transactions", transactions),
    ("portfolio_holdings", holdings),
]:
    orphan_codes = set(df["amfi_code"]) - master_codes
    print(f"{name}: {len(orphan_codes)} amfi_codes not found in fund_master")
    if orphan_codes:
        print(f"  Orphans: {orphan_codes}")
```

Data Quality Summary
------------------------

Completeness:
- All 10 source files loaded successfully with correct datatypes
- Only one null pattern found: "yoy_growth_pct" in "04_monthly_sip_inflows.csv" is null for Jan–Dec 2022 (the first 12 rows). This is not a missing value actually as since there is no same month or prior year to compare against on a year over year growth calculation.
- All other files have zero nulls.

Referential integrity (amfi_code):
- Every amfi_code in nav_history, investor_transactions, and portfolio_holdings matches a real scheme in fund_master and has no orphan/invalid codes anywhere.
- portfolio_holdings only has 34 of the 40 schemes. The missing 6 are all Debt funds — expected, since this file tracks stock holdings, and debt funds hold bonds, not stocks.

Uniqueness / validity:
- nav_history - (46,000 rows): no duplicate AFMI codes and dates pairs and no NAV values ≤ 0
- fund_master - 40 rows = distinct scheme-plan combinations, each with its own AMFI code

Consistency:
- nav_history- covers the same date range (2022-01-03 to 2026-05-29) and same row count (1,150) for all 40 schemes, with dates falling only on weekdays — consistent with mutual funds not publishing NAV on non-trading days

Conclusion:
- the dataset is clean and internally consistent. No imputation or orphan-record removal is needed at ingestion; the only "gaps" found are structurally expected (first-year YoY, debt funds lacking equity holdings) rather than data quality defects.
